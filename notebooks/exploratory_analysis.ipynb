# Saved experiment exploration
This notebook reads artifacts only. It is not part of training, tuning or repair selection.


```python
import json
from pathlib import Path
import pandas as pd
root = Path('../artifacts/core-v4')
results = json.loads((root/'results.json').read_text())
rows = [{'run_id': r['run_id'], 'model': r['architecture'], 'seed': r['seed'], 'method': r['method'], 'budget': r['budget'], 'wga': r['evaluation']['decorrelated_test']['raw']['worst_group_accuracy'], 'matched_accuracy': r['evaluation']['matched_test']['raw']['accuracy']} for r in results]
df = pd.DataFrame(rows)
df
```

```python
df.groupby(['model','method','budget']).wga.agg(['mean','std','count'])
```

```python
import plotly.express as px
px.line(df[df.budget > 0], x='budget', y='wga', color='method', facet_col='model', markers=True)
```
